```python
from rfdetr import RFDETRSegLarge
from rfdetr.datasets.aug_config import AUG_AERIAL
from tqdm.auto import tqdm
import torch

DATASET_PATH = 'datasets'
OUTPUT_PATH = 'output'

model = RFDETRSegLarge()

# Clear GPU memory before training
torch.cuda.empty_cache()

# clear RAM
import gc
gc.collect()

# Augmentation config
aug_config = {
      "HorizontalFlip": {"p": 1.0},
      "VerticalalFlip": {"p": 1.0},
      "RandomRotate90": {"p": 1.0},
}

model.train(
    dataset_dir=DATASET_PATH,
    epochs=150,
    batch_size=16,
    grad_accum_steps=1,
    lr=1e-4,
    early_stopping=False,
    output_dir=OUTPUT_PATH,
    project="sawah-project",
    run="experiment-001",
    resolution=504,
    aug_config=aug_config,
)
```

```python
import supervision as sv
from tqdm import tqdm
from supervision.metrics import MeanAveragePrecision
import os
from PIL import Image
import yaml
import torch

DATATEST_PATH = 'datasets/test'
OUTPUT_PATH = 'prediction'

IMAGES_DIR = os.path.join(DATATEST_PATH, 'images')

# create prediction folder if not exist
if not os.path.exists(OUTPUT_PATH):
    os.mkdir(OUTPUT_PATH)

# Clear GPU memory before inference
torch.cuda.empty_cache()

model_inf = RFDETRSegLarge(pretrain_weights="output/checkpoint_best_total.pth")
model_inf.optimize_for_inference()
    
# Load the Ground Truth Dataset
dataset = sv.DetectionDataset.from_yolo(
    images_directory_path=os.path.join(DATATEST_PATH, 'images'),
    annotations_directory_path=os.path.join(DATATEST_PATH, 'labels'),
    data_yaml_path=os.path.join(DATATEST_PATH, '..', 'data.yaml')
)

targets_list = []
predictions_list = []

# Iterate through the dataset and get model predictions
for image_name, image, target_detections in tqdm(dataset, desc="Processing mAP"):
    image_pil = Image.fromarray(image)

    predictions = model_inf.predict(image_pil, threshold=0.5)
    
    text_scale = sv.calculate_optimal_text_scale(resolution_wh=image_pil.size)
    color = sv.ColorPalette.from_hex([
        "#ffff00", "#ff9b00", "#ff66ff", "#3399ff", "#ff66b2", "#ff8080",
        "#b266ff", "#9999ff", "#66ffff", "#33ff99", "#66ff66", "#99ff00"
    ])
    
    mask_annotator = sv.MaskAnnotator(color=color, opacity=0.3)
    label_annotator = sv.LabelAnnotator(
        color=color,
        text_color=sv.Color.BLACK,
        text_scale=text_scale)
    
    detections_image = image_pil.copy()
    detections_image = mask_annotator.annotate(scene=detections_image, detections=predictions)
    detections_image = label_annotator.annotate(scene=detections_image, detections=predictions)
    
    # save image with overlay annotation
    filename = os.path.basename(image_name)
    output_path = os.path.join(OUTPUT_PATH, filename)
    detections_image.save(output_path)

    targets_list.append(target_detections)
    predictions_list.append(predictions)

# Calculate mAP
map_metric = MeanAveragePrecision()
map_metric.update(predictions_list, targets_list)
metrics = map_metric.compute()

print("\nMean Average Precision metrics:")
print(metrics)
```
